```python
#9.What will happen when you run the following code?


count = 10

def increment():
    count += 1
    print(count)

increment()
```

```python
#Issue: count is defined outside the function, need declare with in function or use global
count = 10

def increment():
    global count
    count += 1
    print(count)

increment()
```
