```python
BMI=int(input("Enter the BMI Index:"))
if(BMI<18.5):
    print("Underweight")
elif(BMI<24.9):
    print("Healthy Weight")
elif(BMI<29.9):
    print("Overweight")
elif(BMI<34.9):
    print("Obesity Class 1")
elif(BMI<39.9):
    print("Obesity Class 2")
else:
    print("Obesity Class 3")   
    
```
