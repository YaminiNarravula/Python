```python
# print Odd Numbers in the list
List=[20,10,16,19,25,1,276,188]
for num in List:
    if((num%2)==1):
        print(num,"is odd")
```

```python
# print Even numbers in the list
List=[20,10,16,19,25,1,276,188]
for num in List:
    if((num%2)==0):
        print(num,"is even")
```
